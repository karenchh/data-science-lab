#### This is everything we need to write a specific cleaning plan:

#### 1-Drop the 117 rows missing coordinates → .dropna()
#### 2-Strip $ and , from price_usd → .str.replace()
#### 3-Convert price_usd from text to float → .astype(float)

```python
df1 = (
    pd.read_csv(path_1)
    .dropna() # drops all the rows that contains at least 1 NaN
        
    # Create clean numeric columns using `.assign()`
    # We use 'lambda x' to refer to the data at this specific step in the chain
    # We overwrite 'price_usd' by accessing the current df (x in our lambda function) and cleaning the string
    
    .assign(
        price_usd = lambda x : x["price_usd"]
            .str.replace("$","",regex = False)
            .str.replace(",","",regex = False)
            .astype(float))
    #.dtypes
    #.isnull().sum()
)
df1.head()
```

### Cleaning data set 2:
#### 1- If we look at the data set we notice that the lat and the lon columns contains NaN (missing values) ==> drop NaN using .dropna().
#### 2- The price is in mxn ==> convert it to usd by dividing it by 19 and add assign it to a new column named price_usd (it is important since later we need to concat the 3 datasets).
#### 3- Drop the old price_mxn column 

```python
df2 = (
    pd.read_csv(path_2)
    .dropna()
    .assign(price_usd = lambda x : x["price_mxn"].div(19)) # covert it from mxn to usd
    .drop(columns = ["price_mxn"])
)
df2.head()
```

```python
path_3 = "./data/mexico-real-estate-3.csv"
```

```python
(
    pd.read_csv(path_3)
      .head()
)
```

```python
# Load the third dataset, remove rows with missing values,
# split the latitude and longitude from the coordinate column,
# extract the state name from the location hierarchy,
# remove the original columns and display the cleaned dataset.

df3 = (
    pd.read_csv(path_3)
        .dropna()
        .assign(lat = lambda x: x['lat-lon']
               .str.split(',' , expand = True)[0].astype(float) ,
                lon = lambda x: x['lat-lon']
               .str.split(',' , expand = True)[1].astype(float) ,
                state = lambda x : x['place_with_parent_names']
                .str.split('|', expand = True)[2]
               )
        .drop(columns=["place_with_parent_names", "lat-lon"])
    # split returns a new data frame with columns 0,1,2...
)
df3.head()
```

```python
# Concat the 3 datasets and reset the index
df = pd.concat([df1,df2,df3], ignore_index = True) 
df.head()
```
